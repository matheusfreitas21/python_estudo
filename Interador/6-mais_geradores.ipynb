```python
#1- Gerador de expressão
quadrados = (x ** 2 for x in range(10))
for quadrado in quadrados:
    print(quadrado)
```

```python
#2 - Interador Infinito
def contador_infinito():
    i = 0
    while True:
        yield i
        i += 1
contador = contador_infinito()

for _ in range(10):
    print(next(contador))
```

```python
#3 - Coleta de Valores com send()

def gerador_soma():
    total = 0
    while True:
        valor = yield total # Recebe um valor e retorn o total acumulado
        if valor is not None:
            total += valor
soma = gerador_soma() #Inicializa o gerador
next(soma)
print(soma.send(10))
print(soma.send(5))
print(soma.send(15))
```

```python
#4 - Tratamento de Exceções
def gerador_excecao():
    try:
        while True:
            valor = yield
            print(f"Valor recebido: {valor}")
    except ValueError:
        print("Valor inválido!")

g = gerador_excecao()
next(g)
g.send(10)
g.throw(ValueError)
```

```python
#5- Cascata de Geradores
def multiplicar_por_dois(iterable):
    for i in iterable:
        yield i * 2

def adicionar_cinco(iterable):
    for i in iterable:
        yield i + 5

numeros = range(5)
resultado = adicionar_cinco(multiplicar_por_dois(numeros))
for r in resultado:
    print(r)
```
